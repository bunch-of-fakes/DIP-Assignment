# Assignment 1.3: Spatial Interpolation

### 1. Import Libraries

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

### 2. Load Image

```python
# Path to input image
image_path = "../assets/input/sample.jpg"

# Read image using OpenCV
image = cv2.imread(image_path)

if image is None:
    raise FileNotFoundError(f"Could not load image: {image_path}")

# Convert BGR to RGB for Matplotlib
image_rgb = cv2.cvtColor(
    image,
    cv2.COLOR_BGR2RGB
)

print("Original image shape:", image_rgb.shape)
```

### 3. Display Original Image

```python
plt.figure(figsize=(7, 5))

plt.imshow(image_rgb)
plt.title("Original Image")
plt.axis("off")

plt.show()
```

### 4. Nearest Neighbor

```python
cv2.INTER_NEAREST

# Scale factor
scale = 2.0

height, width = image.shape[:2]

new_width = int(width * scale)
new_height = int(height * scale)

nearest_image = cv2.resize(
    image,
    (new_width, new_height),
    interpolation=cv2.INTER_NEAREST
)

nearest_rgb = cv2.cvtColor(
    nearest_image,
    cv2.COLOR_BGR2RGB
)

print("Nearest Neighbor shape:", nearest_rgb.shape)
```

### 

### 5. Display Nearest Neighbor

```python
plt.figure(figsize=(8, 6))

plt.imshow(nearest_rgb)
plt.title("Nearest Neighbor Interpolation")
plt.axis("off")

plt.show()
```

### 6. Bilinear Interpolation

```python
cv2.INTER_LINEAR

bilinear_image = cv2.resize(
    image,
    (new_width, new_height),
    interpolation=cv2.INTER_LINEAR
)

bilinear_rgb = cv2.cvtColor(
    bilinear_image,
    cv2.COLOR_BGR2RGB
)

print("Bilinear shape:", bilinear_rgb.shape)
```

### 7. Display Bilinear Interpolation

```python
plt.figure(figsize=(8, 6))

plt.imshow(bilinear_rgb)
plt.title("Bilinear Interpolation")
plt.axis("off")

plt.show()
```

### 8. Bicubic Interpolation

```python
cv2.INTER_CUBIC

bicubic_image = cv2.resize(
    image,
    (new_width, new_height),
    interpolation=cv2.INTER_CUBIC
)

bicubic_rgb = cv2.cvtColor(
    bicubic_image,
    cv2.COLOR_BGR2RGB
)

print("Bicubic shape:", bicubic_rgb.shape)
```

### 

### 9. Display Bicubic Interpolation

```python
plt.figure(figsize=(8, 6))

plt.imshow(bicubic_rgb)
plt.title("Bicubic Interpolation")
plt.axis("off")

plt.show()
```

### 10. Side-by-Side Comparison

```python
fig, axes = plt.subplots(
    2,
    2,
    figsize=(12, 10)
)

# Original
axes[0, 0].imshow(image_rgb)
axes[0, 0].set_title("Original Image")
axes[0, 0].axis("off")

# Nearest Neighbor
axes[0, 1].imshow(nearest_rgb)
axes[0, 1].set_title("Nearest Neighbor")
axes[0, 1].axis("off")

# Bilinear
axes[1, 0].imshow(bilinear_rgb)
axes[1, 0].set_title("Bilinear")
axes[1, 0].axis("off")

# Bicubic
axes[1, 1].imshow(bicubic_rgb)
axes[1, 1].set_title("Bicubic")
axes[1, 1].axis("off")

plt.tight_layout()
plt.show()
```

### 11. Crop a Region

```python
# Select a small region from the original image
crop = image[
    100:200,
    100:200
]

crop_rgb = cv2.cvtColor(
    crop,
    cv2.COLOR_BGR2RGB
)

print("Crop shape:", crop_rgb.shape)
```

### 12. Resize the Same Region

```python
crop_height, crop_width = crop.shape[:2]

zoom_width = crop_width * 5
zoom_height = crop_height * 5

crop_nearest = cv2.resize(
    crop,
    (zoom_width, zoom_height),
    interpolation=cv2.INTER_NEAREST
)

crop_bilinear = cv2.resize(
    crop,
    (zoom_width, zoom_height),
    interpolation=cv2.INTER_LINEAR
)

crop_bicubic = cv2.resize(
    crop,
    (zoom_width, zoom_height),
    interpolation=cv2.INTER_CUBIC
)

crop_nearest = cv2.cvtColor(
    crop_nearest,
    cv2.COLOR_BGR2RGB
)

crop_bilinear = cv2.cvtColor(
    crop_bilinear,
    cv2.COLOR_BGR2RGB
)

crop_bicubic = cv2.cvtColor(
    crop_bicubic,
    cv2.COLOR_BGR2RGB
)
```

### 13. Display Zoomed Comparison

```python
fig, axes = plt.subplots(
    1,
    3,
    figsize=(15, 5)
)

axes[0].imshow(crop_nearest)
axes[0].set_title("Nearest Neighbor")
axes[0].axis("off")

axes[1].imshow(crop_bilinear)
axes[1].set_title("Bilinear")
axes[1].axis("off")

axes[2].imshow(crop_bicubic)
axes[2].set_title("Bicubic")
axes[2].axis("off")

plt.tight_layout()
plt.show()
```
