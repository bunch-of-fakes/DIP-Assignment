# Assignment 2.3: Image Thresholding and Segmentation

### 1. Import Libraries

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

### 2. Load Image

```python
# Path to input image
image_path = "../assets/input/sample.jpg"

# Read image using OpenCV
image = cv2.imread(image_path)

if image is None:
    raise FileNotFoundError(f"Could not load image: {image_path}")

# Convert BGR to RGB for displaying with Matplotlib
image_rgb = cv2.cvtColor(
    image,
    cv2.COLOR_BGR2RGB
)

print("Image shape:", image_rgb.shape)
print("Image data type:", image_rgb.dtype)
```

### 3. Convert to Grayscale

```python
gray_image = cv2.cvtColor(
    image,
    cv2.COLOR_BGR2GRAY
)

print("Grayscale image shape:", gray_image.shape)
```

### 4. Display Original and Grayscale Images

```python
fig, axes = plt.subplots(
    1,
    2,
    figsize=(12, 5)
)

axes[0].imshow(
    image_rgb
)

axes[0].set_title("Original Image")
axes[0].axis("off")


axes[1].imshow(
    gray_image,
    cmap="gray"
)

axes[1].set_title("Grayscale Image")
axes[1].axis("off")

plt.tight_layout()
plt.show()
```

**Binary Thresholding**

The basic thresholding operation is:

$$ s = \begin{cases} 255, & r > T \\ 0, & r \leq T \end{cases} $$

where T is the threshold value.

### 5. Binary Thresholding

```python
threshold_value = 127

_, binary_threshold = cv2.threshold(
    gray_image,
    threshold_value,
    255,
    cv2.THRESH_BINARY
)
```

### 6. Display Binary Threshold

```python
plt.figure(figsize=(7, 5))

plt.imshow(
    binary_threshold,
    cmap="gray"
)

plt.title("Binary Thresholding")
plt.axis("off")

plt.show()
```

**Binary Inverse Thresholding**

In inverse thresholding:

$$ s = \begin{cases} 0, & r > T \\ 255, & r \leq T \end{cases} $$

### 7. Binary Inverse

```python
_, binary_inverse = cv2.threshold(
    gray_image,
    threshold_value,
    255,
    cv2.THRESH_BINARY_INV
)
```

### 8. Display Binary Inverse

```python
plt.figure(figsize=(7, 5))

plt.imshow(
    binary_inverse,
    cmap="gray"
)

plt.title("Binary Inverse Thresholding")
plt.axis("off")

plt.show()
```

**Truncation Thresholding**

For truncation:

$$ s = \begin{cases} T, & r > T \\ r, & r \leq T \end{cases} $$

### 9. Truncation

```python
_, trunc_threshold = cv2.threshold(
    gray_image,
    threshold_value,
    255,
    cv2.THRESH_TRUNC
)
```

### 10. Display Truncation

```python
plt.figure(figsize=(7, 5))

plt.imshow(
    trunc_threshold,
    cmap="gray"
)

plt.title("Truncation Thresholding")
plt.axis("off")

plt.show()
```

**To-Zero Thresholding**

For To-Zero:

$$ s = \begin{cases} r, & r > T \\ 0, & r \leq T \end{cases} $$

### 11. To-Zero Thresholding

```python
_, tozero_threshold = cv2.threshold(
    gray_image,
    threshold_value,
    255,
    cv2.THRESH_TOZERO
)
```

### 12. Display To-Zero

```python
plt.figure(figsize=(7, 5))

plt.imshow(
    tozero_threshold,
    cmap="gray"
)

plt.title("To-Zero Thresholding")
plt.axis("off")

plt.show()
```

**To-Zero Inverse Thresholding**

For To-Zero Inverse:

$$ s = \begin{cases} 0, & r > T \\ r, & r \leq T \end{cases} $$

### 13. To-Zero Inverse

```python
_, tozero_inverse = cv2.threshold(
    gray_image,
    threshold_value,
    255,
    cv2.THRESH_TOZERO_INV
)
```

### 14. Display To-Zero Inverse

```python
plt.figure(figsize=(7, 5))

plt.imshow(
    tozero_inverse,
    cmap="gray"
)

plt.title("To-Zero Inverse Thresholding")
plt.axis("off")

plt.show()
```

### 15. Comparison

```python
fig, axes = plt.subplots(
    2,
    3,
    figsize=(15, 10)
)

axes[0, 0].imshow(
    gray_image,
    cmap="gray"
)
axes[0, 0].set_title("Grayscale")
axes[0, 0].axis("off")


axes[0, 1].imshow(
    binary_threshold,
    cmap="gray"
)
axes[0, 1].set_title("Binary")
axes[0, 1].axis("off")


axes[0, 2].imshow(
    binary_inverse,
    cmap="gray"
)
axes[0, 2].set_title("Binary Inverse")
axes[0, 2].axis("off")


axes[1, 0].imshow(
    trunc_threshold,
    cmap="gray"
)
axes[1, 0].set_title("Truncation")
axes[1, 0].axis("off")


axes[1, 1].imshow(
    tozero_threshold,
    cmap="gray"
)
axes[1, 1].set_title("To-Zero")
axes[1, 1].axis("off")


axes[1, 2].imshow(
    tozero_inverse,
    cmap="gray"
)
axes[1, 2].set_title("To-Zero Inverse")
axes[1, 2].axis("off")

plt.tight_layout()
plt.show()
```
