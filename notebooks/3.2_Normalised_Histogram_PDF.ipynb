# Assignment 3.2: Normalised Histogram (PDF)

### 1. Import Libraries


```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

### 2. Load and Display the Image


```python
image_path = "../assets/input/sample.jpg"

image = cv2.imread(image_path)

if image is None:
    raise FileNotFoundError(f"Image not found: {image_path}")

rgb_image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(rgb_image)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(gray_image, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```

### 3. Generate the Normalised Histogram (PDF)


```python
# Calculate the frequency of each intensity level
histogram = cv2.calcHist([gray_image], [0], None, [256], [0, 256])

# Normalise the histogram to obtain the Probability Density Function (PDF)
total_pixels = gray_image.size
pdf = histogram.flatten() / total_pixels

intensity_levels = np.arange(256)

plt.figure(figsize=(10, 5))
plt.plot(intensity_levels, pdf)
plt.title("Normalised Grayscale Histogram (PDF)")
plt.xlabel("Intensity Level")
plt.ylabel("Probability")
plt.xlim([0, 255])
plt.grid(True, alpha=0.3)
plt.show()

print("Total probability:", np.sum(pdf))
```

### 4. Compare Frequency Histogram and Normalised Histogram


```python
plt.figure(figsize=(14, 5))

plt.subplot(1, 2, 1)
plt.plot(intensity_levels, histogram.flatten())
plt.title("Grayscale Frequency Histogram")
plt.xlabel("Intensity Level")
plt.ylabel("Pixel Count")
plt.xlim([0, 255])
plt.grid(True, alpha=0.3)

plt.subplot(1, 2, 2)
plt.plot(intensity_levels, pdf)
plt.title("Normalised Histogram (PDF)")
plt.xlabel("Intensity Level")
plt.ylabel("Probability")
plt.xlim([0, 255])
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

### 5. Display the PDF as a Bar Chart


```python
plt.figure(figsize=(12, 5))
plt.bar(intensity_levels, pdf, width=1.0)
plt.title("Normalised Histogram (PDF) - Bar Chart")
plt.xlabel("Intensity Level")
plt.ylabel("Probability")
plt.xlim([0, 255])
plt.grid(axis="y", alpha=0.3)
plt.show()
```

### 6. Verify the PDF

```python
print("Image dimensions:", gray_image.shape)
print("Total number of pixels:", total_pixels)
print("Number of intensity levels:", len(pdf))
print("Minimum probability:", np.min(pdf))
print("Maximum probability:", np.max(pdf))
print("Sum of all probabilities:", np.sum(pdf))
```
