# Assignment 2.1: inear Transformations - Brightness and Contrast

### 1. Import Libraries

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

### 2. Load Image

```python
# Path to input image
image_path = "../assets/input/sample.jpg"

# Read image using OpenCV
image = cv2.imread(image_path)

if image is None:
    raise FileNotFoundError(f"Could not load image: {image_path}")

# Convert BGR to RGB for displaying with Matplotlib
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

print("Image shape:", image_rgb.shape)
print("Image data type:", image_rgb.dtype)
```

### 3. Display Original Image

```python
plt.figure(figsize=(7, 5))

plt.imshow(image_rgb)
plt.title("Original Image")
plt.axis("off")

plt.show()
```

### 4. Increase Brightness

```python
# Linear transformation parameters
a = 1.0
b = 50

# s = a*r + b
bright_image = a * image.astype(np.float32) + b

# Keep pixel values within [0, 255]
bright_image = np.clip(
    bright_image,
    0,
    255
).astype(np.uint8)

# Convert BGR to RGB
bright_rgb = cv2.cvtColor(
    bright_image,
    cv2.COLOR_BGR2RGB
)
```

### 5. Display Increased Brightness

```python
plt.figure(figsize=(7, 5))

plt.imshow(bright_rgb)
plt.title("Increased Brightness")
plt.axis("off")

plt.show()
```

### 6. Decrease Brightness

```python
# Linear transformation parameters
a = 1.0
b = -50

# s = a*r + b
dark_image = a * image.astype(np.float32) + b

# Keep pixel values within [0, 255]
dark_image = np.clip(
    dark_image,
    0,
    255
).astype(np.uint8)

# Convert BGR to RGB
dark_rgb = cv2.cvtColor(
    dark_image,
    cv2.COLOR_BGR2RGB
)
```

### 7. Display Decreased Brightness

```python
plt.figure(figsize=(7, 5))

plt.imshow(dark_rgb)
plt.title("Decreased Brightness")
plt.axis("off")

plt.show()
```

### 8. Increase Contrast

```python
# Linear transformation parameters
a = 1.5
b = 0

# s = a*r + b
high_contrast_image = a * image.astype(np.float32) + b

# Keep pixel values within [0, 255]
high_contrast_image = np.clip(
    high_contrast_image,
    0,
    255
).astype(np.uint8)

# Convert BGR to RGB
high_contrast_rgb = cv2.cvtColor(
    high_contrast_image,
    cv2.COLOR_BGR2RGB
)
```

### 9. Display Increased Contrast

```python
plt.figure(figsize=(7, 5))

plt.imshow(high_contrast_rgb)
plt.title("Increased Contrast")
plt.axis("off")

plt.show()
```

### 10. Decrease Contrast

```python
# Linear transformation parameters
a = 0.5
b = 0

# s = a*r + b
low_contrast_image = a * image.astype(np.float32) + b

# Keep pixel values within [0, 255]
low_contrast_image = np.clip(
    low_contrast_image,
    0,
    255
).astype(np.uint8)

# Convert BGR to RGB
low_contrast_rgb = cv2.cvtColor(
    low_contrast_image,
    cv2.COLOR_BGR2RGB
)
```

### 11. Display Decreased Contrast

```python
plt.figure(figsize=(7, 5))

plt.imshow(low_contrast_rgb)
plt.title("Decreased Contrast")
plt.axis("off")

plt.show()
```

### 12. Apply Combined Transformation

```python
# Linear transformation parameters
a = 1.5
b = 30

# s = a*r + b
transformed_image = a * image.astype(np.float32) + b

# Keep pixel values within [0, 255]
transformed_image = np.clip(
    transformed_image,
    0,
    255
).astype(np.uint8)

# Convert BGR to RGB
transformed_rgb = cv2.cvtColor(
    transformed_image,
    cv2.COLOR_BGR2RGB
)
```

### 13. Display Combined Transformation

```python
plt.figure(figsize=(7, 5))

plt.imshow(transformed_rgb)
plt.title("Linear Transformation: s = a·r + b")
plt.axis("off")

plt.show()
```

### 14. Display All Results

```python
fig, axes = plt.subplots(
    2,
    3,
    figsize=(15, 10)
)

# Original
axes[0, 0].imshow(image_rgb)
axes[0, 0].set_title("Original")
axes[0, 0].axis("off")

# Increased brightness
axes[0, 1].imshow(bright_rgb)
axes[0, 1].set_title("Increased Brightness")
axes[0, 1].axis("off")

# Decreased brightness
axes[0, 2].imshow(dark_rgb)
axes[0, 2].set_title("Decreased Brightness")
axes[0, 2].axis("off")

# Increased contrast
axes[1, 0].imshow(high_contrast_rgb)
axes[1, 0].set_title("Increased Contrast")
axes[1, 0].axis("off")

# Decreased contrast
axes[1, 1].imshow(low_contrast_rgb)
axes[1, 1].set_title("Decreased Contrast")
axes[1, 1].axis("off")

# Combined
axes[1, 2].imshow(transformed_rgb)
axes[1, 2].set_title("Brightness + Contrast")
axes[1, 2].axis("off")

plt.tight_layout()
plt.show()
```

### 15. Plot s = a·r + b

```python
# Input intensity values
r = np.arange(256)

# Transformation parameters
a = 1.5
b = 30

# Output intensity
s = a * r + b

# Clip to valid intensity range
s = np.clip(s, 0, 255)

plt.figure(figsize=(7, 5))

plt.plot(r, s)

plt.xlabel("Input Intensity (r)")
plt.ylabel("Output Intensity (s)")
plt.title("Linear Intensity Transformation: s = a·r + b")

plt.xlim(0, 255)
plt.ylim(0, 255)

plt.grid()

plt.show()
```
